#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)

```r
march_madness_tidy <- march_madness |>
    mutate(reg_win_rate = reg_wins / (reg_wins + reg_losses), tourney_win_rate = tourney_wins / (tourney_wins + tourney_losses),
           bid = as_factor(bid)) |>
    select(school, year, bid, reg_win_rate, tourney_win_rate)

head(march_madness_tidy)
```

```r
options(repr.plot.width = 14, repr.plot.height = 8)
mm_reg_hist <- march_madness_tidy |>
    ggplot(aes(x = reg_win_rate)) +
    geom_histogram(binwidth = 0.05, color = "white") +
    labs(x = "Win Rate in Regular Season", y = "Number Of Teams") +
    theme(text = element_text(size = 20))    
mm_reg_hist           
```

```r
options(repr.plot.width = 10, repr.plot.height = 7)
mm_bid_prop <- march_madness_tidy |>
    group_by(bid) |>
    summarize(count = n()) |>
    mutate(proportion = count / sum(count))

mm_bid_plot <- mm_bid_prop |>
    ggplot(aes(x = bid, y = proportion, fill = bid)) +
    geom_col() +
    labs(x = "Bid Type", y = "Proportion of Teams") +
    theme(text = element_text(size = 20), 
          legend.position = "none")
mm_bid_plot
```

Visualization 1: How do the mean tournament win rates compare to the mean regular season win rates for schools that have appeared more than 10 times? This visualization helps with the analysis and research question by providing a base comparison (without much context) of the variables being investigated. This will give us a simple result to show if the question is worth investigating further.

```r
mm_school <- march_madness_tidy |>
    group_by(school) |>
    summarize(mean_reg_rate = mean(reg_win_rate),
              mean_tourney_rate = mean(tourney_win_rate),
              n_appearances = n())

mm_school_plot <- mm_school |>
    filter(n_appearances >= 10) |>
    ggplot(aes(x = mean_reg_rate, y = mean_tourney_rate, size = n_appearances)) +
    geom_point(alpha = 0.5) +
    labs(x = "Mean Regular Season Win Rate (all years)",
         y = "Mean Tournament Win Rate (all years)",
         size = "Tournament\nAppearances") +
    theme(text = element_text(size = 20))
mm_school_plot
```

Second visualization: What does the tournament win rate of the school with the most appearances look like over the years? I wanted to look into this visualization to test if tournament win rates are a reliable variable (since they'll be predicted later in the project). By finding the school with the most appearances, hopefully a good representation can be created.

```r
top_school <- march_madness_tidy |>
    count(school) |>
    slice_max(n, n = 1) |>
    pull(school)

options(repr.plot.width = 12, repr.plot.height = 7)

top_school_plot <- march_madness_tidy |>
    filter(school == top_school) |>
    ggplot(aes(x = year, y = tourney_win_rate)) +
    geom_line() +
    geom_point(size = 3) +
    scale_y_continuous(limits = c(0, 1)) +
    labs(x = "Year",
         y = "Tournament Win Rate",
         title = "Tourney Win Rate by Year: Tennessee") +
    theme(text = element_text(size = 20))
top_school_plot
```
